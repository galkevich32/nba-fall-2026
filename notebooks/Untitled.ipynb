**we need to fix this height column. I'm not even going to work with it but still. I asked Gemini how to fix it**

```python
dates = pd.to_datetime(df_master["Height"], errors="coerce")
df_master["Height"] = (
    dates.dt.month.astype("Int64").astype(str)
    + "-"
    + dates.dt.day.astype("Int64").astype(str)
).where(dates.notna(), df_master["Height"])
```
