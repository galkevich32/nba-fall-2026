**we need to fix this height column. I'm not even going to work with it but still. I asked Gemini how to fix it**

```python
dates = pd.to_datetime(df_master["Height"], errors="coerce")
df_master["Height"] = (
    dates.dt.month.astype("Int64").astype(str)
    + "-"
    + dates.dt.day.astype("Int64").astype(str)
).where(dates.notna(), df_master["Height"])
```

**whoopsie we also have problems with the Season column. it was a good idea to check sample. lets see how many seasons have to be fixed**

```python
season_series = df_master["Season"]
season_series.unique()
```

**lets fix**

```python
df_master["Season"] = (
    df_master["Season"].astype(str).str.replace(r"-01$", "", regex=True)
)
df_master["Season"].unique()
```

**ok now lets pull our College column out**

```python
college_column = df_master["COLLEGE"]
college_column.head()
```

**let's mark all colleges as True and all "Nan" as False**

```python
has_college = college_column.notna()
has_college.head()
```

**lets get counts**

```python
season_college_counts = df_master.groupby(["Season", has_college]).size()
season_college_counts
```

**now let's get the share for each season**

```python
season_college_share = season_college_counts / season_college_counts.groupby(level=0).transform("sum")
season_college_share
```

**cool but lets have it as %**

```python
season_college_pct = (season_college_share * 100).round(2)
season_college_pct
```

**now lets explore drafted vs undrafted angle. we are going to work with the DRAFT ROUND column. lets first see what we have there**

```python
print(df_master["DRAFT ROUND"].unique())
```

**hmm we dont want no NaN here lets see what its all about**

```python
nan_rows = df_master[df_master["DRAFT ROUND"].isna()]
print("Number of nan rows:", len(nan_rows))
print(nan_rows[["Season", "Player", "DRAFT YEAR", "DRAFT ROUND", "DRAFT NUMBER"]].head(15))
```

**ok i manually checked that all these players were undreafted. lets add Undrafted then into each of these 8 rows in columns "DRAFT YEAR", "DRAFT ROUND", "DRAFT NUMBER"**

```python
draft_cols = ["DRAFT YEAR", "DRAFT ROUND", "DRAFT NUMBER"]
mask = df_master["DRAFT ROUND"].isna()

df_master[draft_cols] = df_master[draft_cols].astype("string")
df_master.loc[mask, draft_cols] = "Undrafted"

df_master.loc[[310, 443, 590, 919, 963, 1012, 1024, 1069], ["Player", "Season", "DRAFT YEAR", "DRAFT ROUND", "DRAFT NUMBER"]]
```

**lets double check**

```python
print(df_master["DRAFT ROUND"].unique())
```

**ok seems about right. lets start counting then. first we file each drafted player as drafted**

```python
draft_status = (df_master["DRAFT ROUND"] != "Undrafted").map(
    {True: "Drafted", False: "Undrafted"}
)
```

**and now we count**

```python
season_draft_counts = (
    df_master.groupby(["Season", draft_status])
    .size()
    .unstack(fill_value=0)
    .sort_index()
)

season_draft_counts
```

```python
season_draft_share = season_draft_counts.div(
    season_draft_counts.sum(axis=1), axis=0
)
season_draft_share
```

**and as percentages**

```python
season_draft_pct = (season_draft_share * 100).round(2)
season_draft_pct
```

**ok lets save our findings to csv. we start by renaming columns to avoid duplicate names**

```python
counts_renamed = season_draft_counts.rename(
    columns={"Drafted": "Drafted_Count", "Undrafted": "Undrafted_Count"}
)
pct_renamed = season_draft_pct.rename(
    columns={"Drafted": "Drafted_Pct", "Undrafted": "Undrafted_Pct"}
)
```

**lets combine side-by-side and add total players per season column here**

```python
season_draft_summary = pd.concat([counts_renamed, pct_renamed], axis=1)

season_draft_summary["Total_Players"] = (season_draft_summary["Drafted_Count"] + season_draft_summary["Undrafted_Count"])
```

**first we inspect**

```python
season_draft_summary.sample(10)
```

**then we save. we keep index=True so that Season is saved as our first column**

```python
season_draft_summary.to_csv("../output/season_draft_summary.csv", index=True)
```

**alrighty. now, same for college/non-college. first, we rename our columns, getting rid of True and False, and then we merge counts with pcts**

```python
season_college_summary = pd.concat(
    [
        season_college_counts.unstack(fill_value=0).rename(
            columns={True: "College_Count", False: "No_College_Count"}
        ),
        season_college_pct.unstack(fill_value=0).rename(
            columns={True: "College_Pct", False: "No_College_Pct"}
        ),
    ],
    axis=1,
).sort_index()
```

**lets add Total Players column here as well**

```python
season_college_summary["Total_Players"] = (
    season_college_summary["College_Count"] + season_college_summary["No_College_Count"]
)
```
